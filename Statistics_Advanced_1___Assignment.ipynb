#Theoretical

**Q-1: What is a random variable in probability theory?**
- A random variable is a value that comes from a random process. It connects outcomes of experiments to numbers. For example, tossing a coin: heads = 1, tails = 0. It helps us study probability in a numeric way.

**Q-2: What are the types of random variables?**
- There are two types of random variables: discrete and continuous. Discrete variables take specific, countable values like number of children in a family. Continuous variables can take any value in a range, like height or weight. Discrete values are exact, while continuous values can be infinitely detailed.

**Q-3: Explain the difference between discrete and continuous distributions.**
- Discrete distributions deal with countable outcomes, like rolling a die (1, 2, 3, etc.). Continuous distributions deal with measurements, like height or time, which can take any value within a range. Discrete data has gaps between values, while continuous data is smooth and can have infinite possibilities.

**Q-4: What is a binomial distribution, and how is it used in probability?**
- A binomial distribution shows the probability of getting a certain number of successes in a fixed number of trials. For example, flipping a coin 10 times and counting how many times it lands on heads. It is used when there are only two outcomes (like success/failure) and each trial is independent.

**Q-5: What is the standard normal distribution, and why is it important?**
- The standard normal distribution is a bell-shaped curve with a mean of 0 and a standard deviation of 1. It is important because many statistical tests are based on it. It helps us understand how data is spread and compare different datasets easily.

**Q-6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?**
- The Central Limit Theorem says that if we take many samples from a population, the average of those samples will form a normal distribution, even if the original data is not normal. This is important because it allows us to use normal distribution methods in real-world problems.

**Q-7: What is the significance of confidence intervals in statistical analysis?**
- Confidence intervals give us a range where we believe the true value lies. For example, if we say the average height is 170 cm with a 95% confidence interval of 168–172 cm, we are 95% sure the real average is in that range. It helps us make better decisions with data.

**Q-8: What is the concept of expected value in a probability distribution?**
- Expected value is the average result we expect if we repeat an experiment many times. For example, if we roll a fair die, the expected value is 3.5. It helps us understand the long-term outcome and is useful in making decisions under uncertainty.

#Practical

```python
#Q-9: Write a Python program to generate 1000 random numbers from a normal distribution
#with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy,
#and draw a histogram to visualize the distribution.**

import numpy as np
import matplotlib.pyplot as plt

numbers = np.random.normal(50, 5, 1000)
print("Mean:", np.mean(numbers))
print("Std Dev:", np.std(numbers))
plt.hist(numbers, bins=30, edgecolor='black')
plt.show()
```

```python
#Q-10: You are working as a data analyst for a retail company.
#The company has collected daily sales data for 2 years and wants you to identify the overall sales trend.

#Answer: Using the Central Limit Theorem, we can estimate the average sales and build a confidence interval.
#With 95% confidence, we can say the true mean sales lies in a calculated range.
#This helps management understand the sales trend and make better decisions.


import numpy as np
import scipy.stats as st

daily_sales = [220,245,210,265,230,250,260,275,240,255,235,260,245,250,225,270,265,255,250,260]
mean = np.mean(daily_sales)
stderr = st.sem(daily_sales)
ci = st.t.interval(0.95, len(daily_sales)-1, loc=mean, scale=stderr)
print("Mean:", mean)
print("95% CI:", ci)
```
